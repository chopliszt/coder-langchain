## **Chunking**

Por qué guardamos el texto en los metadatos: la idea es explícitamente no depender de una base relacional aparte para recuperar el texto original — Pinecone permite guardar el contenido dentro del propio vector, en metadata["text"], así una sola consulta trae vector + texto + fuente.

```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = DirectoryLoader("data", glob="*.txt", loader_cls=TextLoader, loader_kwargs={"encoding": "utf-8"})
documentos_crudos = loader.load()

splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(
    chunk_size=600,      # dentro del rango 500-800 tokens sugerido
    chunk_overlap=100,
)

chunks = splitter.split_documents(documentos_crudos)

# Metadata avanzada: fuente, categoría (derivada del nombre de archivo) e índice de chunk
for i, chunk in enumerate(chunks):
    nombre_archivo = os.path.basename(chunk.metadata["source"])
    chunk.metadata["source"] = nombre_archivo
    chunk.metadata["categoria"] = nombre_archivo.replace(".txt", "").replace("_", " ")
    chunk.metadata["chunk_id"] = i

print(f"✂️ Fragmentos generados: {len(chunks)}")
print("Metadata de ejemplo:", chunks[0].metadata)
```

## **Embeddings locales**

```python
from langchain_huggingface import HuggingFaceEmbeddings

EMBEDDING_MODEL = "sentence-transformers/all-MiniLM-L6-v2"
EMBEDDING_DIM = 384  # dimensión de este modelo — el índice de Pinecone debe coincidir exacto

embeddings = HuggingFaceEmbeddings(model_name=EMBEDDING_MODEL)
```

##  **Setup del índice de Pinecone (con verificación de existencia)**

Qué error evita: el "mismatch de dimensiones" que menciona la consigna — fijamos EMBEDDING_DIM una sola vez arriba y lo reutilizamos acá, en vez de hardcodear 1536 a mano (ese número es específico de OpenAI, no de nuestro modelo local).

```python
from pinecone import Pinecone, ServerlessSpec

pc = Pinecone(api_key=os.environ["PINECONE_API_KEY"])

INDEX_NAME = "techcorp-rag-hibrido"
NAMESPACE = "politicas-internas"

indices_existentes = [i["name"] for i in pc.list_indexes()]

if INDEX_NAME not in indices_existentes:
    print(f"🆕 Creando índice '{INDEX_NAME}' (dimensión {EMBEDDING_DIM})...")
    pc.create_index(
        name=INDEX_NAME,
        dimension=EMBEDDING_DIM,
        metric="cosine",
        spec=ServerlessSpec(cloud="aws", region="us-east-1"),
    )
else:
    print(f"♻️ El índice '{INDEX_NAME}' ya existe — no se vuelve a crear")

indice = pc.Index(INDEX_NAME)
print(indice.describe_index_stats())
```

## **Ingesta a Pinecone**

```python
from langchain_pinecone import PineconeVectorStore

vectorstore = PineconeVectorStore.from_documents(
    documents=chunks,
    embedding=embeddings,
    index_name=INDEX_NAME,
    namespace=NAMESPACE,
)

print("📦 Vectores en el namespace:", indice.describe_index_stats()["namespaces"].get(NAMESPACE, {}))
```

## **Retriever léxico (BM25)**

Por qué hace falta además del vectorial: la búsqueda semántica es excelente para parafraseos ("¿cuánto puedo descansar por año?" ≈ "vacaciones"), pero es débil con términos exactos, siglas o nombres propios ("2FA", "MDM"). BM25 busca coincidencias de palabras literales — por eso se combinan.

```python
from langchain_community.retrievers import BM25Retriever

retriever_bm25 = BM25Retriever.from_documents(chunks)
retriever_bm25.k = 5
```

## **Retriever vectorial (Pinecone)**

```python
retriever_vectorial = vectorstore.as_retriever(
    search_kwargs={"k": 5, "namespace": NAMESPACE},
)
```

## **Combinación híbrida (EnsembleRetriever)**

```python
from langchain_classic.retrievers import EnsembleRetriever

retriever_hibrido = EnsembleRetriever(
    retrievers=[retriever_bm25, retriever_vectorial],
    weights=[0.5, 0.5],
)
```

## **Clase RAGSystem**

```python
from typing import List, Dict


class RAGSystem:
    """Encapsula el EnsembleRetriever y expone un método simple para obtener el top-k."""

    def __init__(self, retriever, k: int = 5):
        self.retriever = retriever
        self.k = k

    def obtener_top_k(self, query: str) -> List[Dict]:
        docs = self.retriever.invoke(query)[: self.k]
        return [
            {
                "contenido": d.page_content,
                "fuente": d.metadata.get("source", "desconocida"),
                "categoria": d.metadata.get("categoria", "desconocida"),
            }
            for d in docs
        ]


rag_system = RAGSystem(retriever_hibrido, k=5)
```

## **Prueba rápida**

```python
resultados = rag_system.obtener_top_k("¿Cuántos días de vacaciones tengo con 6 años de antigüedad?")

for i, r in enumerate(resultados, 1):
    print(f"{i}. [{r['fuente']}] {r['contenido'][:120]}...")
```

## **Golden Set de evaluación**

```python
golden_set = [
    {"pregunta": "¿Cuántos días de vacaciones corresponden a un empleado con 6 años de antigüedad?",
     "documento_id_esperado": "politica_vacaciones.txt"},
    {"pregunta": "¿Cuántos días de trabajo remoto por semana tiene el esquema estándar de TechCorp?",
     "documento_id_esperado": "politica_teletrabajo.txt"},
    {"pregunta": "¿Cada cuánto deben renovarse las contraseñas corporativas?",
     "documento_id_esperado": "politica_seguridad_informatica.txt"},
    {"pregunta": "¿Cuánto dura el proceso de onboarding en TechCorp?",
     "documento_id_esperado": "onboarding_nuevos_empleados.txt"},
    {"pregunta": "¿Qué esquema de trabajo tienen las áreas de Soporte Técnico Nivel 1 y Recepción?",
     "documento_id_esperado": "politica_teletrabajo.txt"},
]
```

## **evaluate.py: cálculo de Precision@5 y Recall@5**

```python
def evaluar(rag_system: RAGSystem, golden_set: List[Dict]) -> Dict:
    resultados_por_pregunta = []

    for caso in golden_set:
        top_k = rag_system.obtener_top_k(caso["pregunta"])
        fuentes_recuperadas = [r["fuente"] for r in top_k]

        # Recall@5: ¿el documento esperado aparece entre los recuperados? (0 o 1)
        recall = 1.0 if caso["documento_id_esperado"] in fuentes_recuperadas else 0.0

        # Precision@5: qué % de los 5 recuperados son "útiles" (coinciden con la fuente esperada)
        coincidencias = sum(1 for f in fuentes_recuperadas if f == caso["documento_id_esperado"])
        precision = coincidencias / len(top_k) if top_k else 0.0

        resultados_por_pregunta.append({
            "pregunta": caso["pregunta"],
            "esperado": caso["documento_id_esperado"],
            "recuperados": fuentes_recuperadas,
            "recall@5": recall,
            "precision@5": precision,
        })

    recall_promedio = sum(r["recall@5"] for r in resultados_por_pregunta) / len(resultados_por_pregunta)
    precision_promedio = sum(r["precision@5"] for r in resultados_por_pregunta) / len(resultados_por_pregunta)

    return {
        "detalle": resultados_por_pregunta,
        "recall@5_promedio": recall_promedio,
        "precision@5_promedio": precision_promedio,
    }


reporte = evaluar(rag_system, golden_set)

print("=" * 80)
for r in reporte["detalle"]:
    estado = "✅" if r["recall@5"] == 1.0 else "❌"
    print(f"{estado} {r['pregunta']}")
    print(f"   Esperado: {r['esperado']} | Recuperados: {r['recuperados']}")
    print(f"   Recall@5: {r['recall@5']:.0%} | Precision@5: {r['precision@5']:.0%}\n")

print("=" * 80)
print(f"📊 RECALL@5 PROMEDIO:    {reporte['recall@5_promedio']:.1%}")
print(f"📊 PRECISION@5 PROMEDIO: {reporte['precision@5_promedio']:.1%}")
```

Recall@5 da siempre 0 o 1 por una razón puntual de cómo se armó el Golden Set, no porque la métrica sea binaria por definición: cada pregunta tiene un solo documento_id_esperado. Con un solo documento relevante posible, el cociente solo puede dar 0/1 = 0 o 1/1 = 1 — no hay término medio matemáticamente posible.

## **Prueba final - Consultá el retriever híbrido**

Cada resultado te muestra fuente + categoría + un extracto — así podés ver a simple vista si el retriever híbrido está trayendo el documento correcto, que es justo lo que después mide **evaluate.py** de forma automática.
Si notás que para cierto tipo de preguntas (por ejemplo, con siglas exactas como "2FA") el BM25 aporta más que el vectorial, o al revés, es un buen punto para probar el efecto de cambiar los weights del EnsembleRetriever (probá [0.7, 0.3] o [0.3, 0.7] y volvé a correr algunas preguntas — van a ver el orden de los resultados moverse).
